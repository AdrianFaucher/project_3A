```python
import pandas as pd
import numpy as np
import random
from typing import Optional
G = 4.3009e-9 
```

```python

def velocity_model_1(r, H0, M ):
    Omega_Lambda = 0.67

    
    term1 = H0 * (1.1 + 0.31 * Omega_Lambda) * r
    
    # Check if the term inside sqrt will be negative
    sqrt_term = G * M / r
  
    return term1 - 1.1 * np.sqrt(sqrt_term)
```

```python
import pandas as pd
import numpy as np
import random
from typing import Optional

def generate_galaxy_survey_data(num_galaxies: int, 
                               output_filename: Optional[str] = None,
                               random_seed: Optional[int] = None) -> pd.DataFrame:
    """
    Generate realistic galaxy survey data similar to astronomical catalogs.
    
    Parameters:
    -----------
    num_galaxies : int
        Number of galaxy entries to generate
    output_filename : str, optional
        If provided, saves the data to a CSV file
    random_seed : int, optional
        Random seed for reproducible results
        
    Returns:
    --------
    pd.DataFrame
        Generated galaxy survey data
    """
    
    if random_seed is not None:
        np.random.seed(random_seed)
        random.seed(random_seed)
    
    # Generate base galaxy data
    data = []
    
    for i in range(num_galaxies):
        # Galaxy identification
        pgc_id = f"PGC{random.randint(1000000, 9999999)}"
        

        
        # Distance errors (typically 5-15% of distance)
        
        # Heliocentric velocity (km/s) - roughly follows Hubble law
        H0_minor = 65  # Hubble constant
        M_minor = 10**12
        
        H0_major= 72
        M_major = 10**13

        

        
        # Distance to center of mass system
        dis_center = np.random.uniform(4, 15) 

        # Infall velocities (major and minor components)
        # Major infall velocity (toward center of mass)
        minor_infall = velocity_model_1(dis_center,H0_minor,M_minor) 

        
        # Minor infall velocity (perpendicular component)
        major_infall = velocity_model_1(dis_center,H0_major,M_major) 

        
        # Compile row data
        row = {
            'Unnamed: 0': i,
            'Name': pgc_id,
            'PGC': pgc_id,
            'dis_center_CoM_CenA_M83_0.76':np.absolute(dis_center +np.random.normal(0,0.5)),
            'e_dis_center_min_CoM_CenA_M83_0.76': 0.5,
            'e_dis_center_max_CoM_CenA_M83_0.76': 0.5,
            'major_infall_velocity_CoM_CenA_M83_0.76': major_infall+np.random.normal(0,130),
            'e_major_infall_velocity_min_CoM_CenA_M83_0.76': 130,
            'e_major_infall_velocity_max_CoM_CenA_M83_0.76': 130,
            'minor_infall_velocity_CoM_CenA_M83_0.76': minor_infall+np.random.normal(0,80),
            'e_minor_infall_velocity_min_CoM_CenA_M83_0.76': 80,
            'e_minor_infall_velocity_max_CoM_CenA_M83_0.76': 80,
        }
        
        data.append(row)
    
    # Create DataFrame
    df = pd.DataFrame(data)
    
    # Save to CSV if filename provided
    if output_filename:
        df.to_csv(output_filename, index=False)
        print(f"Generated {num_galaxies} galaxy entries and saved to {output_filename}")
    
    return df

# Example usage and demonstration
if __name__ == "__main__":
    # Generate a small sample for demonstration
    sample_data = generate_galaxy_survey_data(5, random_seed=42)
    print("Sample of generated data:")
    print(sample_data.head())
    
    # Generate larger datasets
    print("\nGenerating larger datasets...")
    
    # Generate 1000 galaxies
    large_dataset = generate_galaxy_survey_data(50000, "generated_data_big.csv", random_seed=888)
    
    # Generate 10000 galaxies
    # very_large_dataset = generate_galaxy_survey_data(10000, "generated_data.csv")
    
    print(f"Large dataset shape: {large_dataset.shape}")
    print("\nColumn names:")
    for i, col in enumerate(large_dataset.columns):
        print(f"{i:2d}: {col}")
```
